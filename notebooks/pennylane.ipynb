```python
import pennylane as qml
```

```python
qml.__version__
```

```python
dev = qml.device('default.qubit', wires = 2)
```

```python
def qc():
    qml.PauliX(wires=0)
    # wires = 0 representa el qubit sobre el que actua el operador
    qml.Hadamard(wires=0)
    return qml.state()
```

```python
qcirc = qml.QNode(qc,dev)
qcirc()

#The result of assembling a circuit function and a device is known in Pennylane as a Quantum Node
```

METODO 2

```python
@qml.qnode(dev)
def qcirc():
    qml.PauliX(wires=0)
    qml.Hadamard(wires=0)
    return qml.state()
```

```python
qcirc()
```

```python
print(qml.draw(qcirc)())
```

```python
dev = qml.device('default.qubit', wires = 1)

@qml.qnode(dev)
#podemos incluir parametros en el circuito
def qcirc1(theta):
    qml.RX(theta,wires = 0)
    return qml.state()
```

```python
#tambien podriamos usar bucles, condicinales...
```

```python
print(qml.draw(qcirc1)(theta = 2))
```

```python
@qml.qnode(dev)
def qcirc2():
    qml.PauliX(wires=0)
    qml.Hadamard(wires=0)
    return qml.probs()
```

```python
qcirc2()
```

```python
@qml.qnode(dev)
def qcirc3():
    qml.PauliX(wires=0)
    qml.Hadamard(wires=0)
    return qml.sample()

new_qnode = qml.set_shots(qcirc3, shots = 16)
```

```python
new_qnode()
```

```python
@qml.set_shots(shots=10)
@qml.qnode(dev)
def qcirc4():
    qml.PauliX(wires=0)
    qml.Hadamard(wires=0)
    return qml.sample()
```

```python
qcirc4()
```
