Podemos usar ahora los Hamiltonianos para crear circuitos QAOA con Pennylane

```python
# elegir p (profundidad)
# inicializar vectores β y γ

# repetir:
#    preparar |ψ(β,γ)> en el ordenador cuántico
#    estimar E(β,γ) = <ψ(β,γ)|C|ψ(β,γ)>
#    actualizar γ y β con un optimizador clásico

# preparar |ψ(β*,γ*)> en el ordenador cuántico
# medir el estado (mas de una vez) o estimar el valor 
# esperado mediante simulacion
```

```python
from pennylane import qaoa
```

```python
# Se declaran los hamiltonianos del problema
# H0 en QAOA toma la forma H0 = X0 + X1 + ... + Xn
H0 = qml.PauliX(0) + qml.PauliX(1)
H1 = 1.0*PauliZ(0)@PauliZ(1)

# dos qubits
wires = range(2)
dev = qml.device('default.qubit', wires=wires)

# numero de capas en el circuito QAOA
p = 2

# se declara una funcion energy para hallar el valor esperado de la energia 
# tras el circuito QAOA
# el estado esta parametrizado mediante (β,γ): angles que se introducen como argumentos
# en la funcion energy


# decorador extiende el comportamiento de energy 
# no devuelve una funcion normal, devuelve un 
# QuantumNode: construye circuitos, ejecuta en hardware/
# simulador y permite gradientes
@qml.qnode(dev)
def energy(angles):
    # se aplica H para obtener el estado inicial: |+>^{\otimes n}
    for wire in wires:
        qml.Hadamard(wire)

    # se aplican las p capas de QAOA
    for layer in range(p):
        qaoa.cost_layer(angles[2*layer+1],H1)
        qaoa.mixer_layer(angles[2*layer],H0)
    
    # valor esperado del hamiltoniano de coste
    return qml.expval(H1)
```

```python
from pennylane import numpy as np
```

```python
optimizer = qml.GradientDescentOptimizer()
steps = 20
angles = np.array([1.0,1.0,1.0,1.0], requires_grad=True)

for i in range(steps):
    angles = optimizer.step(energy, angles)

print(f'optimal angles {angles}')
```

```python
# ahora que hemos obtenido los angulos optimos podemos obtener la solucion al problema

@qml.qnode(dev)
def sample_solutions(angles):
    for wire in wires:
        qml.Hadamard(wire)
    
    for layer in range(p):
        qaoa.cost_layer(angles[2*layer+1],H1)
        qaoa.mixer_layer(angles[2*layer],H0)
    return qml.sample()

print(sample_solutions(angles, shots=5))
```
