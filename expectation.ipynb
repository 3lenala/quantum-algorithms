```python
from qiskit.quantum_info import Statevector
```

INICIALIZACION ESTADOS BASE BINARIA

```python
# De la siguiente forma se puede construir el estado |0>
zero = Statevector([1,0])
print("zero is", zero)
```

```python
# De la siguiente forma se puede construir el estado |0>
one = Statevector([0,1])
print("one is", one)
```

```python
# una forma alternativa mas conveniente de obtener los mismos resultados es inicializando el objeto statevector 
# desde un entero 0 o 1: FROM_INT
# dims se usa para indicar el tamaño del vector de estado dims = 2^n
zero = Statevector.from_int(0, dims=2)
one = Statevector.from_int(1, dims=2)
print("zero is", zero)
print("one is", one)
```

```python
# Ahora podemos construir estados con un mayor numero de qubits mediante producto tensorial con el metodo TENSOR
psi = one.tensor(zero.tensor(zero))
# con esto queremos generar el estado |100>
print("Psi is", psi)
```

```python
#que se puede escribir tambien como
psi1 = one^zero^zero
print("Psi1 is",psi1)
```

```python
# la forma mas rapida de escribir el estado |100> es con FROM_INT dado el numero binario 100 y la dimension
# del estado, en este caso 8

psi2 = Statevector.from_int(4, dims=8)
print("Psi2 is",psi2)
```

INICIALIZACION ESTADODS EN SUPERPOSICION

```python
from numpy import sqrt

ghz = 1/sqrt(2) * (zero^zero^zero) + 1/sqrt(2) * (one^one^one)
print("ghz: ", ghz)
```

Para obtener valores esperados tambien necesitamos Hamiltonianos. Por ahora trabajaremos con productos tensoriales de la puerta Z

```python
from qiskit.quantum_info import Pauli

Z0Z1 = Pauli('ZZI')
print("Z0Z1 es: ", Z0Z1)
print("y su matriz es:")
print(Z0Z1.to_matrix())
```

```python
print("the spars representation of Z0Z1 es la representacion de los coeficientes no nulos, es: ")
print(Z0Z1.to_matrix(sparse=True))
# Los estados que en binario son 2,3,4,5 tienen coeficientes negativos
```

```python
# Los objetos Pauli no se pueden multiplicar por escalares ni sumarse entre ellos
# para poder hacerlo se usa PauliOps

from qiskit.quantum_info import SparsePauliOp
H_cut = SparsePauliOp(Pauli('ZZI')) + SparsePauliOp(Pauli('ZIZ'))
print('Hcut is')
print(H_cut)
```

```python
print(H_cut.to_sparse_list())
```
