```python
import sys
import qiskit
import qiskit_aer

print("Python:", sys.version)
print("qiskit:", qiskit.__version__)
print("qiskit_aer:", qiskit_aer.__version__)
```

```python
from qiskit import QuantumCircuit
from qiskit import transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
qc.save_statevector()
qc.measure_all()
qc.draw()
```

METODO 1 SIMULACION AER: HACER CONTEO

```python
sim = AerSimulator()
compiled = transpile(qc, sim)
result = sim.run(compiled, shots=10000000).result()
print(result.get_counts())
```

METODO 2 SIMULACION AER: OBTENER EL VECTOR DE ESTADO

```python
sim = AerSimulator()
compiled = transpile(qc, sim)
result = sim.run(compiled).result()
sv = result.get_statevector()
print(sv)
```

obtener el vector de estado no es algo que sea factible en un ordenador cuántico real, pero hacer simulaciones tiene esa ventaja
