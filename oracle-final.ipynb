```python
from qiskit.circuit import QuantumCircuit
from qiskit_aer import AerSimulator
```

```python
def simulate(qc: QuantumCircuit):
    sim = AerSimulator()
    compiled = transpile(qc, sim)
    return sim.run(compiled, shots=1).result()
```

```python
def oracle(qc: QuantumCircuit) -> None:
    qc.x(0)
    qc.mcx([0,1,2,3],4)

    qc.x(2)
    qc.mcx([0,1,2,3],4)
    qc.x(0)
    qc.x(2)

    qc.x(3)
    qc.mcx([0,1,2,3],4)
    qc.x(3)
    
    qc.measure(4,0)

    return
```

```python
def apply_oracle(qc: QuantumCircuit) -> int:
    oracle(qc)
    result = simulate(qc)
    return result.get_memory()[0]
```

```python
for i in range(16):
    qc = QuantumCircuit(5, 1)

    # Preparar |i> en q0,q1,q2,q3
    for k in range(4):
        if (i >> k) & 1:
            qc.x(k)

    # q4 empieza en |0>
    oracle_output = apply_oracle(qc)
    print(f'State: {i:04b}; Oracle\'s output: {oracle_output}')
```
