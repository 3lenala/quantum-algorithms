```python
from qiskit.circuit import QuantumCircuit
from qiskit import transpile
from qiskit_aer import AerSimulator
```

```python
def simulate(qc: QuantumCircuit) -> None:
    sim = AerSimulator()
    compiled = transpile(qc, sim)
    result = sim.run(compiled, shots=100).result()
    print(result.get_counts())
```

```python
def oracle(qc: QuantumCircuit) -> None:
    qc.x(0)
    qc.mcx([0,1,2,3],4)

    qc.x(2)
    qc.mcx([0,1,2,3],4)
    qc.x(0)
    qc.x(2)

    qc.x(3)
    qc.mcx([0,1,2,3],4)
    qc.x(3)
    
    qc.measure(4,0)
```

```python
def apply_oracle(qc: QuantumCircuit) -> None:
    oracle(qc)
    simulate(qc)
```

```python
for i in range(16):
    qc = QuantumCircuit(5, 1)

    # Preparar |i> en q0,q1,q2,q3
    for k in range(4):
        if (i >> k) & 1:
            qc.x(k)

    # q4 empieza en |0>
    print(f'State: {i:04b}; Oracle\'s output:')
    apply_oracle(qc)
    
```
