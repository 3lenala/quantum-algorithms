```python
from qiskit.circuit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector
```

```python
def oracle(qc: QuantumCircuit) -> None:
    qc.x(0)
    qc.mcx([0,1,2,3],4)

    qc.x(2)
    qc.mcx([0,1,2,3],4)
    qc.x(0)
    qc.x(2)

    qc.x(3)
    qc.mcx([0,1,2,3],4)
    qc.x(3)

    qc.measure(4,0)

def simulate(qc: QuantumCircuit) -> None:
    sim = AerSimulator()
    compiled = transpile(qc, sim)
    result = sim.run(compiled, shots=10000000).result()
    print(result.get_counts())
```

```python
for i in range(2**4):
    psi = Statevector.from_int(i, dims=2**4).tensor(Statevector.from_int(0, dims=[2]))
    print(f"{i:05b}")
    qc = QuantumCircuit(5,1)
    qc.initialize(psi.data, [0, 1, 2, 3, 4])
    oracle(qc)
    simulate(qc)
```

```python
for i in range(16):
    psi = Statevector.from_int(i, dims=2**4).tensor(Statevector.from_int(0, dims=2))
    qc = QuantumCircuit(5, 1)
    qc.initialize(psi.data, [0,1,2,3,4])
    oracle(qc)
    sv = Statevector.from_instruction(qc.remove_final_measurements(inplace=False))
    probs = sv.probabilities_dict()
    print(i, probs)
```
