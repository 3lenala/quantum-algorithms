QUANTUM FOURIER TRANSFORM FOR 3 QUBITS (GATE BASED COMPUTING)

```python
from qiskit import *
from qiskit_aer import AerSimulator
import numpy as np
```

```python
def QFT(qc):
    n = qc.num_qubits
    for i in range(n):
        qc.h(i)
        for j in range(i+1,n):
            qc.cp(np.pi/2**(j-i),j,i)
    for k in range(n//2):
        qc.swap(k, n-k-1)
    
    qc.save_statevector()
    qc.measure_all()
    return qc
```

```python
qc = QuantumCircuit(4)
qc = QFT(qc)
qc.draw()
```

```python
sim = AerSimulator()
compiled = transpile(qc, sim)
result = sim.run(compiled).result()
sv = result.get_statevector()
print(sv)
```

```python
sim = AerSimulator()
compiled = transpile(qc, sim)
result = sim.run(compiled, shots=10000000).result()
print(result.get_counts())
```
